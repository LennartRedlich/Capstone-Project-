```python
import json, re
import numpy as np
import pandas as pd
from pathlib import Path

ANNOT = Path("linkedin-cvs-annotated.json")
UNLAB = Path("linkedin-cvs-not-annotated.json")

with open(ANNOT, "r", encoding="utf-8") as f:
    profiles_annot = json.load(f)

with open(UNLAB, "r", encoding="utf-8") as f:
    profiles_unlab = json.load(f)

print("annot type:", type(profiles_annot), "len:", len(profiles_annot))
print("unlab type:", type(profiles_unlab), "len:", len(profiles_unlab))

# if list-of-lists, flatten
if isinstance(profiles_annot, list) and len(profiles_annot)>0 and isinstance(profiles_annot[0], list):
    profiles_annot = [x for sub in profiles_annot for x in sub]
if isinstance(profiles_unlab, list) and len(profiles_unlab)>0 and isinstance(profiles_unlab[0], list):
    profiles_unlab = [x for sub in profiles_unlab for x in sub]

print("after flatten: annot len:", len(profiles_annot), "unlab len:", len(profiles_unlab))
print("top keys example:", list(profiles_annot[0].keys()))
```

```python
# DEBUG: inspect one profile keys and nested structure
p0 = profiles_unlab[0]
print("Top keys:", list(p0.keys()))

# print nested keys if there is "linkedin" dict
if isinstance(p0.get("linkedin"), dict):
    print("\nlinkedin keys:", list(p0["linkedin"].keys()))
    for k in ["headline", "summary", "about", "skills", "experience", "positions", "description"]:
        if k in p0["linkedin"]:
            print("FOUND in linkedin:", k, "->", type(p0["linkedin"][k]))
else:
    print("\nNo linkedin dict found. linkedin type:", type(p0.get("linkedin")))
```

```python
def clean_text(x):
    x = "" if x is None else str(x)
    return re.sub(r"\s+", " ", x).strip()

def build_text(row):
    # your schema: position, organization, linkedin, startDate, endDate, status
    pos = clean_text(row.get("position", ""))
    org = clean_text(row.get("organization", ""))
    status = clean_text(row.get("status", ""))
    sd = clean_text(row.get("startDate", ""))
    ed = clean_text(row.get("endDate", ""))

    link = row.get("linkedin", "")
    if isinstance(link, dict):
        link_txt = " ".join([clean_text(v) for v in link.values() if v is not None])
    else:
        link_txt = clean_text(link)

    parts = [pos, org, link_txt, sd, ed, status]
    return " | ".join([p for p in parts if p])

rows = []
for i, r in enumerate(profiles_annot):
    status = str(r.get("status", "")).upper()
    has_active = (status == "ACTIVE")

    rows.append({
        "idx": i,
        "text": build_text(r),
        "domain": r.get("department", None),
        "seniority": r.get("seniority", None),
        "has_active": has_active
    })

df = pd.DataFrame(rows)

df_train = df[
    df["has_active"] &
    df["text"].str.len().gt(0) &
    df["domain"].notna() &
    df["seniority"].notna()
].copy()

print("All:", len(df), "Trainable:", len(df_train))
print("Domains:", df_train["domain"].nunique(), "Seniorities:", df_train["seniority"].nunique())
df_train.head()
```

```python
from sklearn.model_selection import train_test_split
from datasets import Dataset

domain_labels = sorted(df_train["domain"].unique())
seniority_labels = sorted(df_train["seniority"].unique())

domain2id = {l:i for i,l in enumerate(domain_labels)}
id2domain = {i:l for l,i in domain2id.items()}

sen2id = {l:i for i,l in enumerate(seniority_labels)}
id2sen = {i:l for l,i in sen2id.items()}

df_train["domain_id"] = df_train["domain"].map(domain2id)
df_train["seniority_id"] = df_train["seniority"].map(sen2id)

train_df, val_df = train_test_split(
    df_train, test_size=0.2, random_state=42,
    stratify=df_train["domain_id"]
)

ds_dom_train = Dataset.from_pandas(train_df[["text","domain_id"]].rename(columns={"domain_id":"label"}))
ds_dom_val   = Dataset.from_pandas(val_df[["text","domain_id"]].rename(columns={"domain_id":"label"}))

ds_sen_train = Dataset.from_pandas(train_df[["text","seniority_id"]].rename(columns={"seniority_id":"label"}))
ds_sen_val   = Dataset.from_pandas(val_df[["text","seniority_id"]].rename(columns={"seniority_id":"label"}))

print(ds_dom_train, ds_sen_train)
```

```python
import numpy as np
from transformers import AutoTokenizer

MODEL_NAME = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_batch(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

ds_dom_train = ds_dom_train.map(tokenize_batch, batched=True)
ds_dom_val   = ds_dom_val.map(tokenize_batch, batched=True)

ds_sen_train = ds_sen_train.map(tokenize_batch, batched=True)
ds_sen_val   = ds_sen_val.map(tokenize_batch, batched=True)

cols = ["input_ids", "attention_mask", "label"]
ds_dom_train.set_format(type="torch", columns=cols)
ds_dom_val.set_format(type="torch", columns=cols)
ds_sen_train.set_format(type="torch", columns=cols)
ds_sen_val.set_format(type="torch", columns=cols)

print("Tokenized ✔")
```

```python
from sklearn.metrics import accuracy_score, f1_score

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    return {
        "accuracy": accuracy_score(labels, preds),
        "macro_f1": f1_score(labels, preds, average="macro"),
    }
```

```python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer

num_domains = len(domain_labels)

dom_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_domains
)

args_dom = TrainingArguments(
    output_dir="./model_domain",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=3,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    report_to="none",
)

trainer_dom = Trainer(
    model=dom_model,
    args=args_dom,
    train_dataset=ds_dom_train,
    eval_dataset=ds_dom_val,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics,
)

trainer_dom.train()
trainer_dom.evaluate()
```

```python
num_sen = len(seniority_labels)

sen_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_sen
)

args_sen = TrainingArguments(
    output_dir="./model_seniority",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=3,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    report_to="none",
)

trainer_sen = Trainer(
    model=sen_model,
    args=args_sen,
    train_dataset=ds_sen_train,
    eval_dataset=ds_sen_val,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics,
)

trainer_sen.train()
trainer_sen.evaluate()
```

```python
import re
import pandas as pd

def clean_text(x):
    x = "" if x is None else str(x)
    return re.sub(r"\s+", " ", x).strip()

def build_text(row):
    def pick_and_label(label, *keys):
        val = ""
        for k in keys:
            v = row.get(k, None)
            if v is not None and str(v).strip() != "":
                val = str(v)
                break
        return f"{label}: {clean_text(val)}" if val else ""

    parts = []

    pos_val = pick_and_label("Position", "position", "title")
    if pos_val: parts.append(pos_val)

    org_val = pick_and_label("Organization", "organization", "company")
    if org_val: parts.append(org_val)

    dep_val = pick_and_label("Department", "department", "domain")
    if dep_val: parts.append(dep_val)

    # Handle linkedin dict
    link = row.get("linkedin", {})
    if not isinstance(link, dict):
        link = {}

    headline_val = clean_text(link.get("headline", ""))
    if headline_val: parts.append(f"Headline: {headline_val}")

    summary_val = clean_text(link.get("summary", link.get("about", "")))
    if summary_val: parts.append(f"Summary: {summary_val}")

    skills = link.get("skills", "")
    skills_txt = ""
    if isinstance(skills, list):
        skills_txt = " ".join([clean_text(s) for s in skills if clean_text(s)])
    else:
        skills_txt = clean_text(skills)
    if skills_txt: parts.append(f"Skills: {skills_txt}")

    # Enhanced experience extraction
    all_experiences_details = []
    for key in ["experience", "positions", "jobs"]:
        v = link.get(key, None)
        if isinstance(v, list) and len(v) > 0:
            for item in v: # Iterate through all items
                if isinstance(item, dict):
                    item_parts = []
                    title = clean_text(item.get("title", ""))
                    if title: item_parts.append(f"Exp Title: {title}")
                    company = clean_text(item.get("company", ""))
                    if company: item_parts.append(f"Exp Company: {company}")
                    item_sd = clean_text(item.get("startDate", ""))
                    if item_sd: item_parts.append(f"Exp Start: {item_sd}")
                    item_ed = clean_text(item.get("endDate", ""))
                    if item_ed: item_parts.append(f"Exp End: {item_ed}")
                    item_desc = clean_text(item.get("description", ""))
                    if item_desc: item_parts.append(f"Exp Desc: {item_desc}")

                    if item_parts:
                        all_experiences_details.append(" ; ".join(item_parts))
            if all_experiences_details: # Stop after finding the first list and processing it
                break

    if all_experiences_details:
        parts.append(f"Experiences: {' || '.join(all_experiences_details)}") # Use a different separator for multiple experiences

    sd_val = pick_and_label("Start Date", "startDate", "start_date")
    if sd_val: parts.append(sd_val)

    ed_val = pick_and_label("End Date", "endDate", "end_date")
    if ed_val: parts.append(ed_val)

    status_val = pick_and_label("Status", "status", "job_status")
    if status_val: parts.append(status_val)

    return " | ".join([p for p in parts if p])

rows_un = []
for i, r in enumerate(profiles_unlab):
    rows_un.append({
        "idx": i,
        "text": build_text(r),
        "status": str(r.get("status",""))
    })

df_unlab = pd.DataFrame(rows_un)
print(df_unlab.shape)
df_unlab.head()
```

```python
import torch
from torch.utils.data import DataLoader
from datasets import Dataset

def tokenize_only(batch):
    return tokenizer(batch["text"], truncation=True, padding="max_length", max_length=128)

def predict_ids(trainer, texts, batch_size=32):
    ds = Dataset.from_dict({"text": texts})
    ds = ds.map(tokenize_only, batched=True)
    ds.set_format(type="torch", columns=["input_ids","attention_mask"])
    loader = DataLoader(ds, batch_size=batch_size)

    trainer.model.eval()
    device = trainer.model.device
    preds = []

    for batch in loader:
        batch = {k:v.to(device) for k,v in batch.items()}
        with torch.no_grad():
            out = trainer.model(**batch)
            p = out.logits.argmax(dim=-1).cpu().numpy().tolist()
            preds.extend(p)
    return preds

dom_pred_ids = predict_ids(trainer_dom, df_unlab["text"].tolist())
sen_pred_ids = predict_ids(trainer_sen, df_unlab["text"].tolist())

df_unlab["pred_domain"] = [id2domain[i] for i in dom_pred_ids]
df_unlab["pred_seniority"] = [id2sen[i] for i in sen_pred_ids]

df_unlab.to_csv("predictions_finetuned_approach3.csv", index=False)
df_unlab.head()
```

### Key Findings:

1.  **Data Preparation:** The initial data loading, flattening of nested lists, and cleaning for text fields (position, organization, linkedin details) seem effective. The `build_text` function successfully creates a consolidated text representation for each profile.
2.  **Trainable Data:** From an initial `2638` annotated profiles, `623` were deemed 'trainable' after filtering for active status, non-empty text, and existing domain/seniority labels. This suggests that a significant portion of the original annotated dataset might be incomplete or not suitable for direct training.
3.  **Model Performance (Domain Classification):**
    *   The domain classification model achieved an `accuracy` of `0.552` and a `macro_f1` score of `0.0647` on the validation set. The low `macro_f1` suggests that while overall accuracy might be moderate, the model struggles to perform well across all `11` unique domain labels, likely due to class imbalance or difficulty in distinguishing certain domains.
4.  **Model Performance (Seniority Classification):**
    *   The seniority classification model performed slightly better, with an `accuracy` of `0.592` and a `macro_f1` score of `0.3008` on the validation set. This indicates a better ability to classify `6` unique seniority levels, though there's still room for significant improvement.
5.  **Prediction on Unlabeled Data:** The `predict_ids` function and the subsequent mapping back to labels (`id2domain`, `id2sen`) correctly apply the trained models to the unlabeled dataset, generating `pred_domain` and `pred_seniority` columns in `df_unlab`.

### Advantages of this Approach:

1.  **Clear Segmentation:** The problem is effectively broken down into two distinct classification tasks (domain and seniority), allowing for independent model training and evaluation.
2.  **Leveraging Transformers:** Using `distilbert-base-uncased` and the Hugging Face `Trainer` provides a robust, state-of-the-art framework for text classification, potentially offering high performance with sufficient data and fine-tuning.
3.  **Reproducibility:** The use of `random_state=42` for `train_test_split` ensures that the data split is consistent, making the experiments reproducible.
4.  **Comprehensive Text Feature Engineering:** The `build_text` function is quite sophisticated, combining various fields (position, organization, linkedin headline, summary, skills, experience details, dates, status) into a single, rich text input for the models. This provides a good context for classification.
5.  **Evaluation Metrics:** Using `accuracy_score` and `f1_score` (macro) is appropriate for multi-class classification, especially `macro_f1` which accounts for potential class imbalance.

### Limitations of this Approach:

1.  **Limited Data:** The trainable dataset of `623` samples, especially when split into training and validation, might be too small for effective fine-tuning of a large language model like DistilBERT, particularly for `11` domain classes and `6` seniority classes. This is likely a major contributor to the low `macro_f1` scores.
2.  **Class Imbalance:** The low `macro_f1` scores strongly suggest class imbalance within the domain and seniority labels. Without explicit handling (e.g., weighted loss, oversampling/undersampling), the models will struggle to learn from minority classes.
3.  **Hyperparameter Tuning:** The current `TrainingArguments` use default-like parameters. Optimal performance often requires more extensive hyperparameter tuning (e.g., learning rate, batch size, weight decay, number of epochs).
4.  **Lack of Deeper Analysis:** The current output only shows overall evaluation metrics. A per-class breakdown of precision, recall, and F1-score would be very insightful to understand which specific domains/seniorities the models struggle with.
5.  **Generalization Concerns:** Given the relatively small training data and the current performance metrics, the models might not generalize well to unseen, truly diverse profiles.
