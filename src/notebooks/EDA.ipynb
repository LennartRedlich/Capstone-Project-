```python
DATA_PATH = PROJECT_ROOT / "files"
FILE = DATA_PATH / "linkedin-cvs-annotated.json"
FILE.exists()
```

```python
with open(FILE, "r", encoding="utf-8") as f:
    cvs = json.load(f)

type(cvs), len(cvs)
```

Data Type: List
Count of CVs: 609


```python
cvs[0]
```

```python
from collections import Counter

active_job_counts = []

for cv in cvs:
    count = sum(1 for job in cv if job.get("status") == "ACTIVE")
    active_job_counts.append(count)

Counter(active_job_counts)
```

Most individuals have 1 job, second most individuals have zero active jobs and so on...




```python
import pandas as pd

jobs = []
for cv in cvs:
    for job in cv:
        jobs.append(job)

df = pd.DataFrame(jobs)
df.head()



```

```python
df_active = df[df["status"] == "ACTIVE"]
df_active.shape
```

There are 623 jobs with the Status Active currently.

```python
df[["department", "seniority"]].isna().mean()
```
